```python
from langchain_groq import ChatGroq

llm = ChatGroq(
    api_key="ENTER_API_KEY",
    model="llama-3.1-8b-instant",
    temperature=0
)

print("Groq model connected successfully!")
```

```python
import sqlite3

conn = sqlite3.connect("memory.db")

conn.execute("""
CREATE TABLE IF NOT EXISTS facts (
    key TEXT,
    value TEXT
)
""")

conn.commit()

print("Database ready: memory.db")
```

```python
def remember(key, value):
    conn.execute(
        "INSERT INTO facts VALUES (?, ?)",
        (key, value)
    )
    conn.commit()


remember("name", "Tejashwini R K")
remember("favourite_subject", "Science")

print("Saved!")
```

```python
def recall_all():
    rows = conn.execute(
        "SELECT key, value FROM facts"
    ).fetchall()

    return "; ".join(
        f"{k} = {v}" for k, v in rows
    )


memory = recall_all()

print("Recalled memory:", memory)
```

```python
reply = llm.invoke(
    f"Known facts: {memory}. "
    "Greet the user by name and mention their favourite subject."
)

print(reply.content)
```
