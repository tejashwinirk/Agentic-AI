```python
import os

GROQ_API_KEY = os.getenv("GROQ_API_KEY")

config_list = [
    {
        "model": "llama-3.1-8b-instant",
        "api_key": GROQ_API_KEY,
        "base_url": "https://api.groq.com/openai/v1"
    }
]

llm_config = {
    "config_list": config_list,
    "temperature": 0.4
}

print("Model configuration completed!")
```

```python
import autogen

writer = autogen.AssistantAgent(
    name="Writer",
    system_message=(
        "You are a professional blog writer. "
        "Write short, clear and engaging blog posts. "
        "When the Critic gives feedback, rewrite the blog post "
        "to improve it. "
        "Do not explain the changes; provide the revised blog post."
    ),
    llm_config=llm_config,
    human_input_mode="NEVER"
)

critic = autogen.AssistantAgent(
    name="Critic",
    system_message=(
        "You are a strict but helpful blog editor. "
        "Review the Writer's blog post. "
        "Give 2-3 specific improvements. "
        "If the blog post is already good, reply with APPROVED."
    ),
    llm_config=llm_config,
    human_input_mode="NEVER"
)

print("Writer and Critic agents created successfully!")
```

```python
result = critic.initiate_chat(
    writer,
    message="Write a 150-word blog post about the benefits of reading.",
    max_turns=4
)
```

```python
print("FINAL RESULT")
print("=" * 50)
print(result.summary)
```
