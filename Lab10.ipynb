```python
from langchain_groq import ChatGroq

llm = ChatGroq(
    api_key="ENTER_API_KEY",
    model="llama-3.1-8b-instant",
    temperature=0.3
)

print("Groq model connected successfully!")
```

```python
from langchain_community.tools import DuckDuckGoSearchResults

search = DuckDuckGoSearchResults()

def search_agent(topic):
    results = search.run(
        f"latest news about {topic}"
    )
    return results

print("Search Agent created successfully!")
```

```python
def summariser_agent(raw_results):
    prompt = (
        "Summarise the following news search results "
        "into exactly 3 short bullet points. "
        "Use simple and clear language.\n\n"
        + raw_results
    )

    return llm.invoke(prompt).content

print("Summariser Agent created successfully!")
```

```python
def report_agent(topic, summary):
    prompt = (
        f'Create a short Markdown news report titled "News: {topic}".\n\n'
        "Use a main heading and include the following points:\n"
        + summary
    )

    return llm.invoke(prompt).content

print("Report Agent created successfully!")
```

```python
topic = "artificial intelligence"

# Agent 1: Search recent news
raw = search_agent(topic)

# Agent 2: Summarise the search results
summary = summariser_agent(raw)

# Agent 3: Create the final report
report = report_agent(topic, summary)

# Save the report as a Markdown file
with open("news_report.md", "w", encoding="utf-8") as f:
    f.write(report)

print("=" * 60)
print("FINAL NEWS REPORT")
print("=" * 60)
print(report)

print("\n" + "=" * 60)
print("Saved successfully as news_report.md")
print("=" * 60)
```
